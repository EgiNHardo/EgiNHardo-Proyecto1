Paso 3: Definiendo el Nombre de nuestro Video
Ahora, diciéndole al código cómo se llama tu archivo. Pegando esto en una nueva celda. Lo más importante es que se cambie "tu_video.mp4" por el nombre exacto del video que subiste.

```python
# Importando librerías y configurando el nombre del video
import whisper
import os

# --- ¡IMPORTANTE! ---
# ⬇️ Reemplazando "tu_video.mp4" con el nombre de nuestro archivo "OS ESPORTES MAIS POPULARES DO BRASIL !!!.mp4"
nombre_del_video = "OS ESPORTES MAIS POPULARES DO BRASIL !!!.mp4"
# --------------------

ruta_del_video = f"/content/{nombre_del_video}"

# Verificamos si el archivo existe para evitar errores
if os.path.exists(ruta_del_video):
  print(f"✅ Video encontrado: {nombre_del_video}")
else:
  print(f"❌ Error: El archivo '{nombre_del_video}' no existe. Revisa el nombre o súbelo de nuevo.")
```

Paso 4: Transcribiendo el Video a Texto
Este es un paso principal. El código cargará el modelo de inteligencia artificial y procesará nuestro video. Puede tardar varios minutos dependiendo de la duración del video. Pegando en una nueva celda y ejecútala.

```python
# Cargando el modelo y realizar la transcripción
# 'base' es un buen equilibrio entre velocidad y precisión. Otros: tiny, small, medium, large.
print("Cargando el modelo 'base' de Whisper... 🧠")
model = whisper.load_model("base")

print("\nIniciando la transcripción... Esto puede tardar. Verás el texto aparecer aquí abajo. ⏳")
# Whisper detecta el idioma automáticamente y muestra el progreso
result = model.transcribe(ruta_del_video, verbose=True)

print("\n¡Transcripción finalizada! ✨")
```

Paso 5: Guardando el Texto en Formato SRT
El último paso es tomar el texto y los tiempos generados por Whisper y guardarlos en un archivo .srt con el formato correcto. Pegando el código en una última celda y ejecútala.

```python
# Formateando y guardando el resultado como un archivo .srt
import datetime

ruta_srt = f"/content/{os.path.splitext(nombre_del_video)[0]}.srt"

with open(ruta_srt, "w", encoding="utf-8") as srt_file:
    for i, segment in enumerate(result['segments']):
        start_time = str(0) + str(datetime.timedelta(seconds=int(segment['start']))) + ',000'
        end_time = str(0) + str(datetime.timedelta(seconds=int(segment['end']))) + ',000'
        text = segment['text']
        segment_id = i + 1

        srt_file.write(f"{segment_id}\n")
        srt_file.write(f"{start_time} --> {end_time}\n")
        srt_file.write(f"{text.strip()}\n\n")

print(f"✅ ¡Proceso completo! Subtítulos guardados en: {ruta_srt}")
print("\nAhora puedes ir al panel de archivos 📁 de la izquierda, actualizarlo y descargar tu archivo .srt.")
```
