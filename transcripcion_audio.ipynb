Paso 2: Subiendo el Archivo MP3
Este código mostrará un botón para que puedas seleccionar y subir el archivo MP3 desde tu computadora a Colab.

```python
from google.colab import files

print("Por favor, selecciona tu archivo de audio MP3 para subir:")
uploaded = files.upload()

# Guardamos el nombre del archivo para usarlo automáticamente en el siguiente paso
nombre_del_audio = list(uploaded.keys())[0]
print(f'\n✅ ¡Éxito! Se subió el archivo: "{nombre_del_audio}"')
```

Paso 3: Transcribiendo el Audio
Ahora viene la magia. Este bloque de código cargará el modelo de inteligencia artificial y procesará el audio que acabas de subir. La transcripción aparecerá en tiempo real en la salida de la celda.

```python
import whisper

print("Cargando el modelo de Whisper... 🧠")
# Puedes cambiar "base" por "tiny", "small", "medium", o "large" para más precisión
model = whisper.load_model("base")

print(f"Empezando la transcripción de '{nombre_del_audio}'... ⏳")
# El parámetro verbose=True muestra el progreso de la transcripción
result = model.transcribe(nombre_del_audio, verbose=True)

print("\n✨ ¡Transcripción completada!")
```

Paso 4: Guardando y Descargando la Transcripción
Finalmente, este código tomará el texto completo, lo guardará en un archivo .txt y automáticamente iniciará la descarga a tu computadora.

```python
from google.colab import files
import os

# Extraer el texto completo de la transcripción
texto_completo = result["text"]

# Crear un nombre para el archivo de texto de salida
nombre_base = os.path.splitext(nombre_del_audio)[0]
nombre_archivo_txt = f"{nombre_base}_transcripcion.txt"

# Guardar el texto en el archivo
with open(nombre_archivo_txt, "w", encoding="utf-8") as f:
    f.write(texto_completo)

print(f"\n📝 Transcripción guardada en el archivo: '{nombre_archivo_txt}'")

# Descargar el archivo de texto a tu computadora
print("Iniciando la descarga...")
files.download(nombre_archivo_txt)
```
