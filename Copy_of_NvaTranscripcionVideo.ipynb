2. Subir tu video MP4 a Colab

```python
from google.colab import files

print("Selecciona tu archivo MP4...")
uploaded = files.upload()

nombre_video = list(uploaded.keys())[0]
print("Archivo cargado:", nombre_video)
```

3) Recortar los primeros 5 minutos (300 segundos)

⚠️ Completamente reconstruido. Sin shell.

```python
import subprocess

# Archivo de salida
clip_5min = "primeros_5_minutos.mp4"

# Comando ffmpeg para recortar EXACTAMENTE 5 minutos
cmd = [
    "ffmpeg",
    "-y",            # sobrescribir si existe
    "-i", nombre_video,
    "-t", "300",     # 300 segundos = 5 minutos
    "-c:v", "libx264",  # recodificar para asegurar compatibilidad
    "-c:a", "aac",
    clip_5min
]

print("⏳ Recortando los primeros 5 minutos del video...")
subprocess.run(cmd, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
print("✔ Clip creado:", clip_5min)
```

4) Transcribir el clip de 5 minutos

Usamos Whisper con un modelo rápido:

```python
import whisper

print("📌 Cargando modelo (base)...")
model = whisper.load_model("base")   # Más rápido / buena calidad

print("🎤 Transcribiendo...")
resultado = model.transcribe(clip_5min)

texto = resultado["text"]

# Guardar la transcripción
nombre_txt = "primeros_5_minutos_transcripcion.txt"
with open(nombre_txt, "w", encoding="utf-8") as f:
    f.write(texto)

print("✔ Transcripción completa.")
```

5) Descargar el texto

```python
from google.colab import files
files.download(nombre_txt)
```
